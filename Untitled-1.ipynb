 # Creamos el archivo xml y xslt


```python
from lxml import etree

# XML de ejemplo (puede venir de una cadena de texto o un archivo)
xml_data = """
<catalog>
    <book id="1">
        <title>Python Programming</title>
        <price>29.99</price>
    </book>
    <book id="2">
        <title>Learning XPath</title>
        <price>19.99</price>
    </book>
</catalog>
"""


xslt_data = """
<xsl:stylesheet version="1.0" xmlns:xsl="http://www.w3.org/1999/XSL/Transform">
    <!-- Indentación automática para un formato limpio (pretty print) -->
    <xsl:output method="xml" indent="yes" encoding="utf-8"/>

    <!-- Plantilla de identidad: copia por defecto cualquier nodo o atributo -->
    <xsl:template match="@*|node()">
        <xsl:copy>
            <xsl:apply-templates select="@*|node()"/>
        </xsl:copy>
    </xsl:template>

    <!-- Regla específica para el libro con id="1" -->
    <xsl:template match="book[@id='1']">
        <book>
            <!-- Copiamos los atributos existentes (como id="1") -->
            <xsl:copy-of select="@*"/>
            
            <!-- Copiamos el título actual -->
            <xsl:apply-templates select="title"/>
            
            <!-- Modificamos el precio -->
            <price>24.99</price>
            
            <!-- Añadimos el nuevo nodo <author> -->
            <author>Guido van Rossum</author>
        </book>
    </xsl:template>
    <xsl:template match="title"/>
</xsl:stylesheet>
"""
```

# PASO 1: Parsear el XML en un árbol de objetos

```python
root = etree.fromstring(xml_data.strip())

xslt_dom = etree.fromstring(xslt_data.strip())
```

Usamos la consulta xpath buscnado los títulos, el libro con id=1 en el caso de que exista

```python


titles = root.xpath('//book/title/text()')
print("Títulos encontrados:", titles)

# Buscar un libro específico por su atributo id usando XPath
specific_book = root.xpath("//book[@id='1']")
if specific_book:
    print(f"Libro con ID 1 encontrado: {specific_book[0].find('title').text}")

```

# Paso 2 modificamos el arbol de forma programática

```python


# --- PASO 3: Modificar el contenido del árbol ---
print("\n--- 3. Modificando el XML ---")

if specific_book:
    book_node = specific_book[0]
    
    # Modificar el valor de un nodo existente (el precio)
    price_node = book_node.find('price')
    price_node.text = '24.99'
    print("Precio actualizado a 24.99")
    
    # Añadir un nuevo elemento (ej. el autor)
    author_node = etree.Element('author')
    author_node.text = 'Guido van Rossum'
    book_node.append(author_node)
    print("Nuevo nodo <author> añadido.")


```

# Paso 3 crwamos un tranform con el XSLT y lo aplicamos a nuestro xml

```python
transform = etree.XSLT(xslt_dom)
nuevoXml=transform(root)
```

Mostramos el resultado de la solucion programática

```python
resultado_xml_XSLT = str(nuevoXml)

print("\n--- 4. XML Modificado y Reformateado ---")
print(etree.tostring(root, pretty_print=True, encoding='utf-8').decode('utf-8'))

```
